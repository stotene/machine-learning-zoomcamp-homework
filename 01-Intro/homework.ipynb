```python
# Select the first 7 cars from Asia
asia_cars = df[df["origin"] == "Asia"].head(7)

# Create the X matrix using vehicle_weight and model_year
X = asia_cars[["vehicle_weight", "model_year"]].values

# Create the target vector
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# Calculate X^T X
XTX = X.T @ X

# Calculate the inverse of X^T X
XTX_inv = np.linalg.inv(XTX)

# Calculate the weights
w = XTX_inv @ X.T @ y

# Calculate the sum of the weights
w_sum = w.sum()

print("Weights:", w)
print("Sum of weights:", w_sum)
```
